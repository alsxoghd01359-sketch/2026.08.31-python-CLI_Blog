```python
from datetime import datetime, timedelta

print("== CLI 게시판 실행 ==")
last_article_id = 1
articles = [{"id": 1, "title": "aa", "contens": "ss", "time": "2026-08-31 12:12:12"}]  # 날짜 테스트를 하기위한 예시 데이터
tday = datetime.today() + timedelta(hours=9)
tday = tday.strftime('%Y-%m-%d %H:%M:%S')
tday = str(tday)
while True:
  command = input("명령어 ) ").strip()

  # 게시글 작성 기능
  if command == "article write":
    last_article_id += 1
    id = last_article_id
    title = input("제목: ")
    contens = input("내용: ")
    now_time = datetime.today() + timedelta(hours=9)
    now_time = now_time.strftime('%Y-%m-%d %H:%M:%S')

    article = {"id": id, "title": title, "contens": contens, "time": now_time}

    articles.append(article)
    print("{}번 글이 생성되었습니다.".format(last_article_id))


  # 게시글 목록 출력 기능
  elif command == "article list":
    print("==========================================================")
    print("{0:<7}/{1:^13}/{2:^15}/{3:^15}".format("번호", "제목", "내용", "작성일"))
    for a in reversed(articles):
      a_time = a['time']
      if tday[:10] == a_time[:10]:
        print("  {0:<7}/{1:^13}/{2:^15}/{3:^15}".format(a['id'], a['title'], a['contens'], a_time[11:]))
      else:
        print("  {0:<7}/{1:^13}/{2:^15}/{3:^15}".format(a['id'], a['title'], a['contens'], a_time[:10]))


    print("==========================================================")

    last_article_id = id
    # 게시글 목록 기능에도 추가할 기능이 있음.
    # 목록에서도 작성 시간을 추가해야하는데, 작성한 날짜에 따라 변화가 있어야함.
    # 오늘 날짜에 작성한 글이 아닐때만 날짜가 나오고, 오늘 작성한 글이라면 시간만 나오는 형식

    # 지금까지 출력만 신경썼지 시간을 자세히 보지 않아서 몰랐는데 시간 차이가 났었다. 9시간정도
    # 찾아보니 timedelta 모듈로 시간을 더할수 있다고 한다.

  # 게시글 삭제 기능
  elif "article delete" in command:
    delete_cmd = command[14:].strip()
    if delete_cmd.isdigit() == False:
      print("잘못된 명령어가 입력되었습니다.")
      continue
    delete_id = int(delete_cmd)

    delete_bool = True

    new_articles = []
    delete_articles =[]

    for a in articles:
      if a['id'] == delete_id:
        delete_articles.append(a)
        delete_bool = False
        print("{}번 글이 삭제되었습니다.".format(delete_id))
        break

    for a in articles:
      if a not in delete_articles:
        new_articles.append(a)

    articles = new_articles

    if delete_bool:
      print("{}번 글은 존재하지 않습니다.".format(delete_id))


  elif "article edit" in command:
    edit_cmd = command[12:].strip()
    if edit_cmd.isdigit() == False:
      print("잘못된 명령어가 입력되었습니다.")
      continue
    edit_id = int(edit_cmd)

    edit_bool = True
    edit_articles = []

    for a in articles:
      if a['id'] == edit_id:
        edit_articles.append(a)
        edit_index = articles.index(a)
        edit_bool = False
        break

    if edit_bool:
      print("{}번 글은 존재하지 않습니다.".format(edit_id))
      continue

    print("기존 제목 : {}".format(edit_articles[0]['title']))
    print("기존 내용 : {}".format(edit_articles[0]['contens']))
    new_title = input("새 제목 : ")
    new_contens = input("새 내용 : ")

    edit_articles[0]['title'] = new_title
    edit_articles[0]['contens'] = new_contens
    articles[edit_index] = edit_articles[0]

  elif "article detail" in command:
    detail_cmd = command[14:].strip()
    if detail_cmd.isdigit() == False:
      print("잘못된 명령어가 입력되었습니다.")
      continue
    detail_id = int(detail_cmd)
    detail_bool = True

    for a in articles:
      if a['id'] == detail_id:
        print("번호 : {}".format(a['id']))
        print("날짜 : {}".format(a['time']))
        print("제목 : {}".format(a['title']))
        print("내용 : {}".format(a['contens']))
        detail_bool = False
        break

    if detail_bool:
      print("{}번 글은 존재하지 않습니다.".format(detail_id))

  elif command == "exit":
    print("== CLI 게시판 종료 ==")
    break
  else:
    print("잘못된 명령어가 입력되었습니다.")
```

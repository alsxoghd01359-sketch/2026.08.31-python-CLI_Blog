```python
from datetime import datetime, timedelta

class Article:
  def __init__(self, id, m_id):
    self.id = id
    self.title = input("제목: ")
    self.contens = input("내용: ")
    self.time = datetime.today() + timedelta(hours=9)
    self.time = self.time.strftime('%Y-%m-%d %H:%M:%S')
    self.m_id = m_id

class Member:
  # 회원의 고유번호를 나타내는 id를 추가하여 게시글의 삭제 및 수정을 다룰 예정
  def __init__(self, id, name, user_id, pw):
    self.id = id
    self.name = name
    self.user_id = user_id
    self.pw = pw

# def makeTestData():
#   return [
#       Article(1,"aa1","ss1","2026-08-31 12:12:12"),
#       Article(2,"aa2","ss2","2026-09-01 12:12:12"),
#       Article(3,"aa3","ss3","2026-09-01 12:12:12")
#       ]

def id_value(cmd):
  cmdBits = command.split(" ")

  if len(cmdBits) < 3:
    print("명령어의 길이가 짧습니다")
    return False

  if not cmdBits[2].isdigit():
    print("번호가 정수형태가 아닙니다.")
    return False
  editedId = int(cmdBits[2])

  return editedId

def id_in_article(id, a_list):

  for a in a_list:
    if a.id == id:
      return a
  return None

print("== CLI 게시판 실행 ==")
last_article_id = 0
m_id = 1
login_m_id = 0
# 로그인을 성공했다면 고유번호를 부여받은 상태로 기능을 사용할 수 있음. 글 작성에서 사용할 예정
login_token = 0
# 로그인 토큰을 사용해 로그인을 성공했다면 회원 전용 기능을 사용할 수 있게 함
articles = []
members = []

while True:
  command = input("명령어 ) ").strip()
  tday = datetime.today() + timedelta(hours=9)
  tday = tday.strftime('%Y-%m-%d %H:%M:%S')
  tday = str(tday)

  # 게시글 작성 기능
  if command == "article write":
    if login_token == 1:
      last_article_id += 1
      article = Article(last_article_id, login_m_id)

      print("login_m_id = {} , article.m_id = {} , member.id = {}".format(login_m_id, article.m_id, member.id))

      articles.append(article)
      print("{}번 글이 생성되었습니다.".format(last_article_id))
    if login_token != 1:
      print("회원만 글 작성이 가능합니다.")


  # 게시글 목록 출력 기능
  elif command == "article list":
    print("===============================================================================")
    print("{0:<7}/{1:^13}/{2:^15}/{3:^15}".format("번호", "제목", "내용", "작성일"))
    for a in reversed(articles):
      if tday[:10] == a.time[:10]:
        print("  {0:<7}/{1:^13}/{2:^15}/{3:^15}".format(a.id, a.title, a.contens, a.time[11:]))
      else:
        print("  {0:<7}/{1:^13}/{2:^15}/{3:^15}".format(a.id, a.title, a.contens, a.time[11:]))
    print("===============================================================================")


  # 게시글 삭제 기능
  elif command.startswith("article delete"):
    # login_m_id과 글에 등록된 m_id가 일치할때만 기능 사용이 가능
    if login_m_id == article.m_id:
      print("login_m_id = {} , article.m_id = {} , member.id = {}".format(login_m_id, article.m_id, member.id))
      delete_Id = id_value(command)
      if delete_Id == False:
        continue

      article = id_in_article(delete_Id, articles)
      if article == None:
        print("{}번 글은 존재하지 않습니다.".format(delete_Id))
        continue
      articles.remove(article)
      print("{}번 글이 삭제되었습니다.".format(delete_Id))
    else:
      print("다른 회원의 글을 수정/삭제는 불가능 합니다.")


  # 게시글 수정
  elif command.startswith("article edit"):
    # login_m_id과 글에 등록된 m_id가 일치할때만 기능 사용이 가능
    if login_m_id == article.m_id:
      edit_Id = id_value(command)
      if edit_Id == False:
        continue

      article = id_in_article(edit_Id, articles)
      if article == None:
        print("{}번 글은 존재하지 않습니다.".format(edit_Id))
        continue

      print("기존 제목 : {}".format(article.title))
      print("기존 내용 : {}".format(article.contens))
      article.title = input("새 제목 : ")
      article.contens = input("새 내용 : ")
    else:
      print("다른 회원의 글을 수정/삭제는 불가능 합니다.")


  # 게시글 상세보기
  elif command.startswith("article detail"):
    detail_Id = id_value(command)
    if detail_Id == False:
      continue

    article = id_in_article(detail_Id, articles)
    if article == None:
      print("{}번 글은 존재하지 않습니다.".format(detail_Id))
      continue
    print("번호 : {}".format(article.id))
    print("날짜 : {}".format(article.time))
    print("제목 : {}".format(article.title))
    print("내용 : {}".format(article.contens))

  elif command == "member join":
    name = input("이름 : ")
    while True:
      # 2개 이상의 회원정보가 있을 경우 중복 확인이 제대로 안됨.
      # 로그인 로직 수정 완
      user_id = input("아이디 : ")
      i = 0
      for m in members:
        if m.user_id == user_id:
          i = 1
          print("중복된 아이디입니다. 다시 입력해주세요")
          break
      if i == 0:
        break
    while True:
      pw = input("비밀번호 : ")
      pw_check = input("한번 더 입력해주세요 : ")
      if pw != pw_check:
        print("비밀번호가 다릅니다.")
      elif pw == pw_check:
        print("비밀번호가 일치합니다.")
        break
    member = Member(m_id, name, user_id, pw)
    members.append(member)

    m_id += 1

  elif command == "member login":
    login_id = input("아이디 : ")
    login_pw = input("비밀번호 : ")
    for m in members:
      if login_id == m.user_id and login_pw == m.pw:
        print("로그인 성공")
        login_token = 1
        login_m_id = m.id
        print(login_m_id)
        break
        # 로그인을 성공한다면 토큰과 고유번호를 받음
        # 토큰이랑 고유번호를 하나로 통합한다면?
        # 결국 토큰은 0이 아닌 숫자이기만 하면 기능을 전부 사용할 수 있으니 상관이 없을텐데
        # 혹시 모르니 두 개로 나눠놓자

        # 현재 회원이 2명 이상인 상황에서 로그인에 성공했을때,
        # 로그인 성공과 실패가 둘 다 출력됨. 내가 생각하기엔 로그인 조건문에서 문제가 생기는거 같음
        # 아 break를 안써서 그런건가?
        # 오류가 뜨는 이유를 알아냄. 지금 조건문의 방식이 입력받은 login_id와 login_pw를 members 리스트를 순회하면서 찾아내는 것이기 때문에
        # 첫번째에 있는 회원으로 로그인할때는 로그인에 성공하고 break로 탈출하기 때문에 아무런 문제가 없지만.
        # 그 이후에 있는 회원으로 로그인할때는 순회하는 중에 입력받은 login_id와 login_pw와 가져온 member의 인스턴스가 아니라면 로그인 실패를 출력하는거임.

        # 그럼 어떻게 고침?
        # 1. 가장 간단한 방법. 로그인 실패를 출력하지않는거임
        # 문제점: 로그인에 성공했는지 실패했는지 출력하지 않는 경우가 생기기에 로그인이 됐다고 착각할 수 있음.
        # 애초에 로그인에 실패한걸 출력안하는게 말이되나?

        # 2. 마지막 member 인스턴스까지 순회한 뒤 회원정보가 없다면 로그인 실패를 출력.
        # 이게 가장 좋은 방법 아닐까? 그럼 가져온 인스턴스가 어떻게 마지막 인스턴스인지 알아내냐가 관건.
        # 현재 회원가입을 할때마다 회원 고유번호를 기록하기 위해 m_id를 1씩 증가시키며 등록중. 결국 이게 회원이 몇명이냐 or 몇번 회원까지 있냐를 알아낼 수 있음
        # 그럼 현재 m_id를 불러와 member의 아이디가 같고, login_id와 login_pw가 등록되어 있지 않을때만 출력한다면?
        # 일단 로그인 실패가 안뜨는 것은 확인됨. 그럼 두 번째 문제였던 login_m_id와 m_id가 일치하지 않을때 지워지는 문제를 해결해야함.
      elif m_id == login_id and login_id != m.user_id and login_pw != m.pw:
        print("로그인 실패")

  elif command == "member logout":
    # login_token으로 로그인 상태를 검사함
    if login_token == 0:
      print("로그인 상태가 아닙니다.")
      continue
    check = input("정말로 로그아웃 하시겠습니까? (yes or no) : ")
    if check == "yes":
      login_token = 0
      login_m_id = 0
      # 로그아웃을 하면 login_token과 login_m_id을 다시 가져옴
      print("로그아웃 되셨습니다.")
      continue
    elif check == "no":
      continue
    else:
      print("잘못된 명령어가 입력되었습니다.")

  elif command == "member list":
    for m in members:
      print(" {} / {} / {} / {} ".format(m.id, m.name, m.user_id, m.pw))
  # 회원 정보가 제대로 들어갔는지 확인 용도

  elif command == "exit":
    print("== CLI 게시판b 종료 ==")
    break
  else:
    print("잘못된 명령어가 입력되었습니다.")
```

```python
import datetime  # 게시글 상세 보기 페이지에서 날짜를 표시하기 위해 필요

print("== CLI 게시판 실행 ==")
last_article_id = 0
articles = []

while True:
  command = input("명령어 ) ").strip()

  # 게시글 작성 기능
  if command == "article write":
    last_article_id += 1
    id = last_article_id
    title = input("제목: ")
    contens = input("내용: ")
    now_time = datetime.datetime.now()
    now_time = now_time.strftime('%Y-%m-%d %H:%M:%S')

    article = {"id": id, "title": title, "contens": contens, "time": now_time}
    # now_time 값을 추가하게 된다면 수정하는 것도 변경해야하나?
    # 기존의 수정 방식이 아예 덮어 씌우는 방식이였어서
    # 수정해야할 키값만 불러와 수정하는 것으로 변경

    articles.append(article)
    print("{}번 글이 생성되었습니다.".format(last_article_id))


  # 게시글 목록 출력 기능
  elif command == "article list":
    print("==========================================")
    print("{0:<7}/{1:^13}/{2:^15}".format("번호", "제목", "내용"))
    for a in reversed(articles):
      print("  {0:<7}/{1:^13}/{2:^15}".format(a['id'], a['title'], a['contens']))
    print("==========================================")

    last_article_id = id

  # 게시글 삭제 기능
  elif "article delete" in command:
    delete_cmd = command[14:].strip()
    if delete_cmd.isdigit() == False:
      print("잘못된 명령어가 입력되었습니다.")
      continue
    delete_id = int(delete_cmd)
    # 인덱스가 앞으로 당겨지면서 다른 게시글이 삭제되는 경우는 해결
    # 하지만 특정 번호의 게시글이 없는 경우의 텍스트를 출력하는데 문제가 생김
    # 지금 상태로는 조건문이 참이 아닌 경우마다 존재하지 않는다고 출력 중
    # continue와 함께 반복문 밖으로 뺄 경우 게시글을 삭제했을때는 정상적으로 출력이 되지만
    # 특정 번호의 게시글이 없는 경우의 텍스트가 출력되지 않음
    # 삭제와 텍스트와 게시글이 존재하지 않는 다는 텍스트의 위치를 바꿔 시도. -> 이 경우엔 인덱스 0부터 찾다가 한번씩 존재하지 않는다고 출력.

    # delete_bool 변수를 추가하고 지워졌다면 False로 값을 바꾸고
    # 지워지지 않았다면 True 유지, delete_bool이 True 일때만 출력 > 성공
    # 근데 수정해야됨. 기존에는 del을 사용해서 인덱스로 값을 지웠는데 값을 지운상태에서
    # 새로운 게시글이 저장되면 인덱스와 id값이 매치되지않아 지워지지않음
    # 탄막 게임하고 유사한 경우

    # 결국 탄막 게임때 해결했던 방식하고 똑같이 id값이 같은 딕셔너리를 delete_articles에 넣어두고
    # delete_articles에 없는 딕셔너리만 new_articles에 넣은뒤 기존 리스트에 다시 옮기는 방법으로 해결함.

    # 수정 후 생각해보니 지금 명령어에서 숫자를 가져오는 로직이 한자리 수일때만 가능함
    # 결국 이 조건문에 온다는 것은 "article delete"은 무조건 입력 한다는 것. 때문에 14자리 뒤에 오는 숫자만 걸러낸다면
    # 몇자리 숫자가 오든 상관없음. 다만 14자리 이후 띄어쓰기를 안하고 입력한다면 ex) "article deletes 3" s가 걸러지면서 오류가 남.
    # 이건 isdigit() 함수를 사용하여 참이라면 그대로 진행, 거짓이라면 명령어를 재입력하라고 continue하면됨.
    delete_bool = True

    new_articles = []
    delete_articles =[]

    for a in articles:
      if a['id'] == delete_id:
        delete_articles.append(a)
        delete_bool = False
        print("{}번 글이 삭제되었습니다.".format(delete_id))
        break

    for a in articles:
      if a not in delete_articles:
        new_articles.append(a)

    articles = new_articles

    if delete_bool:
      print("{}번 글은 존재하지 않습니다.".format(delete_id))


  elif "article edit" in command:
    edit_cmd = command[12:].strip()
    if edit_cmd.isdigit() == False:
      print("잘못된 명령어가 입력되었습니다.")
      continue
    edit_id = int(edit_cmd)

    edit_bool = True
    edit_articles = []

    for a in articles:
      if a['id'] == edit_id:
        edit_articles.append(a)
        edit_index = articles.index(a)
        edit_bool = False
        break

    if edit_bool:
      print("{}번 글은 존재하지 않습니다.".format(edit_id))
      continue

    print("기존 제목 : {}".format(edit_articles[0]['title']))
    print("기존 내용 : {}".format(edit_articles[0]['contens']))
    new_title = input("새 제목 : ")
    new_contens = input("새 내용 : ")

    edit_articles[0]['title'] = new_title
    edit_articles[0]['contens'] = new_contens
    articles[edit_index] = edit_articles[0]

    # 수정기능도 새로운 리스트를 만들고 수정할 게시글 값을 가져온다음 이번에는 다시 넣는게 아닌
    # id에 맞는 값으로 변경해주면 되지 않을까
    # 인덱스 값까지 추출하는게 확실한 듯

    # 기존 게시글의 제목과 내용이 안나옴. 반복문에서 새 리스트에 제대로 들어갔는지 테스트(하는김에 인덱스 값까지)
    # 중괄호를 안넣어놨었음 실수 ㅎ

    # 아무런 게시글도 지우지 않은 상태에선 아무런 일 없이 수정가능
    # 지우거나 중간에 추가되는 게시글도 문제없이 수정가능

    # 존재하지 않는 게시글을 불러올 경우 텍스트 출력은 delete 기능과 동일한 로직 사용

  elif "article detail" in command:
    detail_cmd = command[14:].strip()
    if detail_cmd.isdigit() == False:
      print("잘못된 명령어가 입력되었습니다.")
      continue
    detail_id = int(detail_cmd)
    detail_bool = True

    for a in articles:
      if a['id'] == detail_id:
        print("번호 : {}".format(a['id']))
        print("날짜 : {}".format(a['time']))
        print("제목 : {}".format(a['title']))
        print("내용 : {}".format(a['contens']))
        detail_bool = False
        break

    if detail_bool:
      print("{}번 글은 존재하지 않습니다.".format(detail_id))

  elif command == "exit":
    print("== CLI 게시판 종료 ==")
    break
  else:
    print("잘못된 명령어가 입력되었습니다.")
```

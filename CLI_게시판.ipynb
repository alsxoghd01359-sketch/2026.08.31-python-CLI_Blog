```python
from datetime import datetime, timedelta

class Article:
  def __init__(self, id, mb_id):
    self.id = id
    self.title = input("제목: ")
    self.contens = input("내용: ")
    self.time = datetime.today() + timedelta(hours=9)
    self.time = self.time.strftime('%Y-%m-%d %H:%M:%S')
    self.mb_id = mb_id

class Member:
  # 회원의 고유번호를 나타내는 id를 추가하여 게시글의 삭제 및 수정을 다룰 예정
  # 상당히 애를 먹었다 로직 자체는 금방 만들었지만 정확히 어떤 데이터를 member.uq_id를 사용했을때 정확히 어떤 인스턴스의 값이 나오는지 몰랐다.
  # 계속 맨 끝에 있는 인스턴스의 id를 가져다 썼고 그거 때문에 권한인식이 제대로 안됐던 것같다.
  def __init__(self, id, name, user_id, pw):
    self.uq_id = id
    self.name = name
    self.user_id = user_id
    self.pw = pw

# def makeTestData():
#   return [
#       Article(1,"aa1","ss1","2026-08-31 12:12:12"),
#       Article(2,"aa2","ss2","2026-09-01 12:12:12"),
#       Article(3,"aa3","ss3","2026-09-01 12:12:12")
#       ]

def id_value(cmd):
  cmdBits = command.split(" ")

  if len(cmdBits) < 3:
    print("명령어의 길이가 짧습니다")
    return False

  if not cmdBits[2].isdigit():
    print("번호가 정수형태가 아닙니다.")
    return False
  editedId = int(cmdBits[2])

  return editedId

def id_in_article(id, a_list):

  for a in a_list:
    if a.id == id:
      return a
  return None

def id_in_member(user_id, m_list):
  for member in m_list:
    if member.user_id == user_id:
      return member
  return None

print("== CLI 게시판 실행 ==")
last_article_id = 0
last_member_id = 0
login_m = None
# 고유번호가 아닌 로그인 된 인스턴스 자체를 가져다 사용함
articles = []
members = []

while True:
  command = input("명령어 ) ").strip()
  tday = datetime.today() + timedelta(hours=9)
  tday = tday.strftime('%Y-%m-%d %H:%M:%S')
  tday = str(tday)

  # 회원 가입 기능
  if command == "member join":
    last_member_id += 1
    name = input("이름 : ")
    while True:
      user_id = input("아이디 : ")

      member = id_in_member(user_id, members)
      if member != None:
        print("중복된 아이디가 있습니다.")
        continue
      break

    while True:
      pw = input("비밀번호 : ")
      pw_check = input("한번 더 입력해주세요 : ")
      if pw != pw_check:
        print("비밀번호가 다릅니다.")
      elif pw == pw_check:
        print("비밀번호가 일치합니다.")
        print("회원가입 성공.")
        break
    member = Member(last_member_id, name, user_id, pw)
    members.append(member)

  # 로그인 기능
  elif command == "member login":
    if login_m != None:
      print("이미 로그인 중 입니다.")
      continue
    login_id = input("아이디 : ")
    login_pw = input("비밀번호 : ")

    member = id_in_member(login_id, members)

    if member == None:
      print("회원 정보가 존재하지 않습니다.")
      continue

    if member.pw != login_pw:
      print("비밀번호가 틀렸습니다.")
      continue
    print("로그인 성공")
    login_m = member


    # if len(members) == 0:
    #   print("회원 정보가 존재하지 않습니다.")
    #   continue
    # for m in members:
    #   if login_id == m.user_id and login_pw == m.pw:
    #     print("로그인 성공")
    #     login_m = m
    #     break
    #   elif last_member_id == login_id and login_id != m.user_id and login_pw != m.pw:
    #     print("회원 정보가 존재하지 않습니다.")

  elif command == "member logout":
    if login_m == None:
      print("로그인 상태가 아닙니다.")
      continue
    check = input("정말로 로그아웃 하시겠습니까? (yes or no) : ")
    if check == "yes":
      login_m = None
      # 로그아웃을 하면 변수를 다시 None 값으로 만든다.
      print("로그아웃 되셨습니다.")
      continue
    elif check == "no":
      continue
    else:
      print("잘못된 명령어가 입력되었습니다.")

  elif command == "member list":
    for m in members:
      print(" {} / {} / {} / {} ".format(m.uq_id, m.name, m.user_id, m.pw))
  # 회원 정보가 제대로 들어갔는지 확인 용도

  # 게시글 작성 기능
  elif command == "article write":
    if login_m != None:
      last_article_id += 1

      article = Article(last_article_id, login_m.uq_id)

      articles.append(article)
      print("{}번 글이 생성되었습니다.".format(last_article_id))
    elif login_m == None:
      print("회원만 글 작성이 가능합니다.")


  # 게시글 목록 출력 기능
  elif command == "article list":
    print("===============================================================================")
    print("{0:<7}/{1:^13}/{2:^15}/{3:^15}".format("번호", "제목", "내용", "작성일"))
    for a in reversed(articles):
      if tday[:10] == a.time[:10]:
        print("  {0:<7}/{1:^13}/{2:^15}/{3:^15}".format(a.id, a.title, a.contens, a.time[11:]))
      else:
        print("  {0:<7}/{1:^13}/{2:^15}/{3:^15}".format(a.id, a.title, a.contens, a.time[11:]))
    print("===============================================================================")


  # 게시글 삭제 기능
  elif command.startswith("article delete"):
    if login_m == None:
      print("로그인이 필요합니다.")
      continue

    delete_Id = id_value(command)
    if delete_Id == False:
      continue

    article = id_in_article(delete_Id, articles)

    # login_last_member_id과 글에 등록된 last_member_id가 일치할때만 기능 사용이 가능
    if article.mb_id != login_m.uq_id:
      print("다른 회원의 글은 삭제할 수 없습니다.")
      continue

    if article == None:
      print("{}번 글은 존재하지 않습니다.".format(delete_Id))
      continue
    articles.remove(article)
    print("{}번 글이 삭제되었습니다.".format(delete_Id))

  # 게시글 수정
  elif command.startswith("article edit"):
    if login_m == None:
      print("로그인이 필요합니다.")
      continue

    edit_Id = id_value(command)
    if edit_Id == False:
      continue

    article = id_in_article(edit_Id, articles)

    if article.mb_id != login_m.uq_id:
      print("다른 회원의 글은 수정할 수 없습니다.")
      continue

    if article == None:
      print("{}번 글은 존재하지 않습니다.".format(edit_Id))
      continue

    print("기존 제목 : {}".format(article.title))
    print("기존 내용 : {}".format(article.contens))
    article.title = input("새 제목 : ")
    article.contens = input("새 내용 : ")


  # 게시글 상세보기
  elif command.startswith("article detail"):
    detail_Id = id_value(command)
    if detail_Id == False:
      continue

    article = id_in_article(detail_Id, articles)
    if article == None:
      print("{}번 글은 존재하지 않습니다.".format(detail_Id))
      continue
    print("번호 : {}".format(article.id))
    print("날짜 : {}".format(article.time))
    print("제목 : {}".format(article.title))
    print("내용 : {}".format(article.contens))

  elif command == "exit":
    print("== CLI 게시판b 종료 ==")
    break
  else:
    print("잘못된 명령어가 입력되었습니다.")
```

```python
print("== CLI 게시판 실행 ==")
last_article_number = 0
articles = []

while True:
  command = input("명령어 ) ").strip()

  # 게시글 작성 기능
  if command == "article write":
    last_article_number += 1
    id = last_article_number
    title = input("제목: ")
    contens = input("내용: ")

    article = {"id": id, "title": title, "contens": contens}

    articles.append(article)

    print("{}번 글이 생성되었습니다.".format(last_article_number))


  # 게시글 목록 출력 기능
  elif command == "article list":
    print("==========================================")
    print("{0:<7}/{1:^13}/{2:^15}".format("번호", "제목", "내용"))
    for a in reversed(articles):
      print("  {0:<7}/{1:^13}/{2:^15}".format(a['id'], a['title'], a['contens']))
    print("==========================================")

    last_article_number = id

  # 게시글 삭제 기능
  elif "article delete" in command:
    delete_number = int(command[len(command) - 1:])

    if delete_number > len(articles) or delete_number <= 0:
      print("{}번 글은 존재하지 않습니다.".format(delete_number))
      continue

    del articles[delete_number - 1]
    print("{}번 글이 삭제 되었습니다.".format(delete_number))

  elif command == "article edit":
    pass

  elif command == "article detail":
    pass

  elif command == "exit":
    print("== CLI 게시판 종료 ==")
    break
  else:
    print("잘못된 명령어가 입력되었습니다.")
```

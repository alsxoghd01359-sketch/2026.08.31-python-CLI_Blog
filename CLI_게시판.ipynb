```python
from datetime import datetime, timedelta

class Article:
  def __init__(self, id, m_id):
    self.id = id
    self.title = input("제목: ")
    self.contens = input("내용: ")
    self.time = datetime.today() + timedelta(hours=9)
    self.time = self.time.strftime('%Y-%m-%d %H:%M:%S')
    self.m_id = m_id

class Member:
  # 회원의 고유번호를 나타내는 id를 추가하여 게시글의 삭제 및 수정을 다룰 예정
  def __init__(self, id, name, user_id, pw):
    self.id = id
    self.name = name
    self.user_id = user_id
    self.pw = pw

# def makeTestData():
#   return [
#       Article(1,"aa1","ss1","2026-08-31 12:12:12"),
#       Article(2,"aa2","ss2","2026-09-01 12:12:12"),
#       Article(3,"aa3","ss3","2026-09-01 12:12:12")
#       ]

def id_value(cmd):
  cmdBits = command.split(" ")

  if len(cmdBits) < 3:
    print("명령어의 길이가 짧습니다")
    return False

  if not cmdBits[2].isdigit():
    print("번호가 정수형태가 아닙니다.")
    return False
  editedId = int(cmdBits[2])

  return editedId

def id_in_article(id, a_list):
  for a in a_list:
    if a.id == id:
      return a
  return None

print("== CLI 게시판 실행 ==")
last_article_id = 0
m_id = 1
login_m_id = 0
# 로그인을 성공했다면 고유번호를 부여받은 상태로 기능을 사용할 수 있음. 글 작성에서 사용할 예정
login_token = 0
# 로그인 토큰을 사용해 로그인을 성공했다면 회원 전용 기능을 사용할 수 있게 함
articles = []
members = []

while True:
  command = input("명령어 ) ").strip()
  tday = datetime.today() + timedelta(hours=9)
  tday = tday.strftime('%Y-%m-%d %H:%M:%S')
  tday = str(tday)

  # 게시글 작성 기능
  if command == "article write":
    if login_token == 1:
      last_article_id += 1
      article = Article(last_article_id, login_m_id)

      articles.append(article)
      print("{}번 글이 생성되었습니다.".format(last_article_id))
    if login_token != 1:
      print("회원만 글 작성이 가능합니다.")


  # 게시글 목록 출력 기능
  elif command == "article list":
    print("==========================================================")
    print("{0:<7}/{1:^13}/{2:^15}/{3:^15}".format("번호", "제목", "내용", "작성일"))
    for a in reversed(articles):
      if tday[:10] == a.time[:10]:
        print("  {0:<7}/{1:^13}/{2:^15}/{3:^15}".format(a.id, a.title, a.contens, a.time[11:]))
      else:
        print("  {0:<7}/{1:^13}/{2:^15}/{3:^15}".format(a.id, a.title, a.contens, a.time[11:]))
    print("==========================================================")


  # 게시글 삭제 기능
  elif command.startswith("article delete"):
    delete_Id = id_value(command)
    if delete_Id == False:
      continue

    article = id_in_article(delete_Id, articles)
    if article == None:
      print("{}번 글은 존재하지 않습니다.".format(delete_Id))
      continue
    articles.remove(article)
    print("{}번 글이 삭제되었습니다.".format(delete_Id))


  # 게시글 수정
  elif command.startswith("article edit"):
    edit_Id = id_value(command)
    if edit_Id == False:
      continue

    article = id_in_article(edit_Id, articles)
    if article == None:
      print("{}번 글은 존재하지 않습니다.".format(edit_Id))
      continue

    print("기존 제목 : {}".format(article.title))
    print("기존 내용 : {}".format(article.contens))
    article.title = input("새 제목 : ")
    article.contens = input("새 내용 : ")


  # 게시글 상세보기
  elif command.startswith("article detail"):
    detail_Id = id_value(command)
    if detail_Id == False:
      continue

    article = id_in_article(detail_Id, articles)
    if article == None:
      print("{}번 글은 존재하지 않습니다.".format(detail_Id))
      continue
    print("번호 : {}".format(article.id))
    print("날짜 : {}".format(article.time))
    print("제목 : {}".format(article.title))
    print("내용 : {}".format(article.contens))

  elif command == "member join":
    name = input("이름 : ")
    while True:
      # 중복된 아이디라면 계속 반복해서 입력할 수 있도록 무한루프
      # 만약 탈출 명령어가 따로 필요하다면 추가 예정
      user_id = input("아이디 : ")
      i = 0
      if len(members) == 0:
        break
      for m in members:
        if m.user_id == user_id:
          print("중복된 아이디입니다. 다시 입력해주세요")
          break
        # 중복 아이디가 아니라면 i에 1을 넣어 반복문을 탈출
        i = 1
      if i == 1:
        break
    while True:
      # 비밀번호가 다르다면 계속 반복해서 입력할 수 있도록 무한루프
      # 만약 탈출 명령어가 따로 필요하다면 추가 예정
      pw = input("비밀번호 : ")
      pw_check = input("한번 더 입력해주세요 : ")
      if pw != pw_check:
        print("비밀번호가 다릅니다.")
      elif pw == pw_check:
        print("비밀번호가 일치합니다.")
        break
    member = Member(m_id, name, user_id, pw)
    members.append(member)

    m_id += 1

  elif command == "member login":
    login_id = input("아이디 : ")
    login_pw = input("비밀번호 : ")
    for m in members:
      if login_id == m.user_id and login_pw == m.pw:
        print("로그인 성공")
        login_token = 1
        login_m_id = m.id
        # 로그인을 성공한다면 토큰과 고유번호를 받음
        # 토큰이랑 고유번호를 하나로 통합한다면?
        # 결국 토큰은 0이 아닌 숫자이기만 하면 기능을 전부 사용할 수 있으니 상관이 없을텐데
        # 혹시 모르니 두 개로 나눠놓자
      else:
        print("로그인 실패")

  # elif command == "member list":
  #   for m in members:
  #     print(" {} / {} / {} / {} ".format(m.id, m.name, m.user_id, m.pw))
  # 회원 정보가 제대로 들어갔는지 확인 용도

  elif command == "exit":
    print("== CLI 게시판 종료 ==")
    break
  else:
    print("잘못된 명령어가 입력되었습니다.")
```

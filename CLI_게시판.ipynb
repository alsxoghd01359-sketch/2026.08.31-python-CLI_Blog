```python
print("== CLI 게시판 실행 ==")
last_article_id = 0
articles = []

while True:
  command = input("명령어 ) ").strip()

  # 게시글 작성 기능
  if command == "article write":
    last_article_id += 1
    id = last_article_id
    title = input("제목: ")
    contens = input("내용: ")

    article = {"id": id, "title": title, "contens": contens}

    articles.append(article)
    print("{}번 글이 생성되었습니다.".format(last_article_id))


  # 게시글 목록 출력 기능
  elif command == "article list":
    print("==========================================")
    print("{0:<7}/{1:^13}/{2:^15}".format("번호", "제목", "내용"))
    for a in reversed(articles):
      print("  {0:<7}/{1:^13}/{2:^15}".format(a['id'], a['title'], a['contens']))
    print("==========================================")

    last_article_id = id

  # 게시글 삭제 기능
  elif "article delete" in command:
    delete_id = int(command[len(command) - 1:])

    # 인덱스가 앞으로 당겨지면서 다른 게시글이 삭제되는 경우는 해결
    # 하지만 특정 번호의 게시글이 없는 경우의 텍스트를 출력하는데 문제가 생김
    # 지금 상태로는 조건문이 참이 아닌 경우마다 존재하지 않는다고 출력 중
    # continue와 함께 반복문 밖으로 뺄 경우 게시글을 삭제했을때는 정상적으로 출력이 되지만
    # 특정 번호의 게시글이 없는 경우의 텍스트가 출력되지 않음
    # 삭제와 텍스트와 게시글이 존재하지 않는 다는 텍스트의 위치를 바꿔 시도. -> 이 경우엔 찾다가 존재하지 않는다고 출력.

    # 반복문을 빠져나왔을때 d의 값이 초기화 되지않는다면 가능한 방법이 있음
    # 삭제가 된 값에서 break를 사용해 반복문에서 탈출하고 그 값을 가지고 조건문에서 탈출하면됨
    # 삭제가 안됐다면 결국 d의 값으로 아무것도 한게 없는게 되므로 그대로 존재하지않는다고 출력하면됨 -> 이것도 안됨

    # delete_bool 변수를 추가하고 지워졌다면 True로 값을 바꾸고
    # 지워지지 않았다면 False로 유지, delete_bool이 False 일때만 출력 > 성공
    # 근데 약간 수정해야됨. 기존에는 del을 사용해서 인덱스로 값을 지웠는데 이러니까 값을 지운상태에서 새로운 게시글이 올라오면 인덱스와 id값이 매치되지않아 지워지지않음
    # 탄막 게임하고 유사한 경우

    # 결국 탄막 게임때 해결했던 방식하고 똑같이 id값이 같은 딕셔너리를 delete_articles에 넣어두고
    # delete_articles에 없는 딕셔너리만 new_articles에 넣은뒤 기존 리스트에 다시 옮기는 방법으로 해결함.

    # 수정 후 생각해보니 지금 명령어에서 숫자를 가져오는 로직이 한자리 수일때만 가능함
    #
    delete_bool = True

    new_articles = []
    delete_articles =[]

    for a in articles:
      if a['id'] == delete_id:
        delete_articles.append(a)
        delete_bool = False
        print("{}번 글이 삭제되었습니다.".format(delete_id))
        break


    for a in articles:
      if a not in delete_articles:
        new_articles.append(a)

    articles = new_articles

    if delete_bool:
      print("{}번 글은 존재하지 않습니다.".format(delete_id))


  elif "article edit" in command:
    edit_id = int(command[len(command) - 1:])

    for e in range(len(articles) - 1):
      if edit_id == articles[e]['id']:
        update_title = input("제목: ")
        update_contens = input("내용: ")
        articles[edit_id - 1]['title'] = update_title
        articles[edit_id - 1]['contens'] = update_contens
        continue
      else:
        print("{}번 글은 존재하지 않습니다.".format(edit_id))

  # elif "article detail" in command:
  #   detail_id = int(command[len(command) - 1:])

  #   if detail_id > len(articles) or detail_id <= 0:
  #     print("{}번 글은 존재하지 않습니다.".format(detail_id))
  #     continue

  #   print("==========================================")
  #   print(  "제목: ".format(articles[detail_id - 1, 1]))
  #   print("==========================================")


  elif command == "exit":
    print("== CLI 게시판 종료 ==")
    break
  else:
    print("잘못된 명령어가 입력되었습니다.")
```

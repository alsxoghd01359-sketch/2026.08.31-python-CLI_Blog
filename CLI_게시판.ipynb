```python
from datetime import datetime, timedelta

class Article:
  def __init__(self, id):
    self.id = id
    self.title = input("제목: ")
    self.contens = input("내용: ")
    self.time = datetime.today() + timedelta(hours=9)
    self.time = self.time.strftime('%Y-%m-%d %H:%M:%S')

  def print_article(self, day):
    self.tday = day
    if self.tday[:10] == self.time[:10]:
      print("  {0:<7}/{1:^13}/{2:^15}/{3:^15}".format(self.id, self.title, self.contens, self.time[11:]))
    else:
      print("  {0:<7}/{1:^13}/{2:^15}/{3:^15}".format(self.id, self.title, self.contens, self.time[:10]))


# def makeTestData():
#   return [
#       {"id": 1, "title": "aa", "contens": "ss", "time": "2026-08-31 12:12:12"},
#       {"id": 2, "title": "abb", "contens": "ss", "time": "2026-08-31 12:13:12"},
#       {"id": 3, "title": "acc", "contens": "saqs", "time": "2026-08-31 12:14:12"}
#       ]

def id_value(cmd):
  cmdBits = command.split(" ")

  if len(cmdBits) < 3:
    print("명령어의 길이가 짧습니다")
    return False

  if not cmdBits[2].isdigit():
    print("번호가 정수형태가 아닙니다.")
    return False
  editedId = int(cmdBits[2])

  return editedId

def id_in_article(id, a_list):
  for a in a_list:
    if a['id'] == id:
      return a
  return None

print("== CLI 게시판 실행 ==")
last_article_id = 0
articles = []

while True:
  command = input("명령어 ) ").strip()
  tday = datetime.today() + timedelta(hours=9)
  tday = tday.strftime('%Y-%m-%d %H:%M:%S')
  tday = str(tday)
  # 게시글 작성 기능
  if command == "article write":
    last_article_id += 1
    article = Article(last_article_id)

    articles.append(article)
    print("{}번 글이 생성되었습니다.".format(last_article_id))


  # 게시글 목록 출력 기능
  elif command == "article list":
    print("==========================================================")
    print("{0:<7}/{1:^13}/{2:^15}/{3:^15}".format("번호", "제목", "내용", "작성일"))
    for article in reversed(articles):
      article.print_article(tday)
    print("==========================================================")


  # 게시글 삭제 기능
  elif command.startswith("article delete"):
    delete_Id = id_value(command)
    if delete_Id == False:
      continue

    article = id_in_article(delete_Id, articles)
    if article == None:
      print("{}번 글은 존재하지 않습니다.".format(delete_Id))
      continue
    articles.remove(article)
    print("{}번 글이 삭제되었습니다.".format(delete_Id))


  # 게시글 수정
  elif command.startswith("article edit"):
    edit_Id = id_value(command)
    if edit_Id == False:
      continue

    article = id_in_article(edit_Id, articles)
    if article == None:
      print("{}번 글은 존재하지 않습니다.".format(edit_Id))
      continue

    print("기존 제목 : {}".format(article['title']))
    print("기존 내용 : {}".format(article['contens']))
    article['title'] = input("새 제목 : ")
    article['contens'] = input("새 내용 : ")


  # 게시글 상세보기
  elif command.startswith("article detail"):
    detail_Id = id_value(command)
    if detail_Id == False:
      continue

    article = id_in_article(delete_Id, articles)
    if article == None:
      print("{}번 글은 존재하지 않습니다.".format(delete_Id))
      continue
    print("번호 : {}".format(article['id']))
    print("날짜 : {}".format(article['time']))
    print("제목 : {}".format(article['title']))
    print("내용 : {}".format(article['contens']))


  elif command == "exit":
    print("== CLI 게시판 종료 ==")
    break
  else:
    print("잘못된 명령어가 입력되었습니다.")
```

```python
print("== CLI 게시판 실행 ==")
i = 0
title_list = []
contens_list = []

while True:
  command = input("명령어 ) ")

  if command == "article write":

    title = input("제목: ")
    contens = input("내용: ")

    title_list.append(title)
    contens_list.append(contens)
    i += 1
    print("{}번 글이 생성되었습니다.".format(i))

  elif command == "article list":
    print("=====================================================")
    print("{0:<7}/     {1:<15}/{2:>15}".format("번호", "제목", "내용"))
    while i >= 1:
      i -= 1
      print("  {0:<7}/     {1:<15}/{2:>15}".format(i + 1, title_list[i], contens_list[i]))
    print("=====================================================")
  elif command == "exit":
    print("== CLI 게시판 종료 ==")
    break
  else:
    print("잘못된 명령어가 입력되었습니다.")
```
